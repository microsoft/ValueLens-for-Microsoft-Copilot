# Agent 365 - Registry Ingester (Fabric) — recommended default

> **Status: supported (app-only, unattended) — this is the recommended default path** for landing the
> Agent 365 registry into the Lakehouse. As of **PAX `purview-v1.11.11`** and the current
> Microsoft Graph docs, the *Agent 365 catalog* endpoint supports **application permissions** and
> **`/v1.0`**, so this notebook runs **headless on a schedule** with a service principal — no
> interactive sign-in. (It was previously a delegated-only, interactive PREVIEW.)

## What this does

Pulls the tenant's **Agent 365 agent catalogue** (declarative agents, plugins, etc. - including
their data-access **capabilities/permissions**: OneDrive/SharePoint read, Graph connector, code
interpreter, image generation, uploaded files) into the Lakehouse Delta table `dbo.agents_365`
(the table the dashboard reads), keyed on `Title ID` (the `T_`-prefixed titleId).

## Requirements & caveats (read before using)

| Item | Detail |
|---|---|
| **Auth** | **App-only / client credentials** (service principal). Runs unattended. No user sign-in. |
| **Permissions** | **Application** permissions `CopilotPackages.Read.All` **+** `Application.Read.All` **+** `User.Read.All`, **admin-consented**. `User.Read.All` is new - it backs creator resolution tier 1. |

**Relationship to the export lander:** `Copilot_Agent365_Lander.ipynb` (admin-center **export CSV**)

`dbo.agents_365` table, so the shipped pipeline runs this notebook **first** and the Lander **only if
this one fails** (activity `Run_Agent365_CSV_Fallback`). If you schedule them yourself, never run both.

## Incremental detail pull (`Detail As Of`)

The catalog **list** call is cheap and runs in full on every run. The per-agent **detail** call
(the only source of Bot Id, usage, sharing and capabilities) is the slow part, so it runs only
for agents that are:

- new, or have no entry in the cache table `dbo.agents_365_detail_cache`;
- changed - `lastModifiedDateTime` **differs** from the cached value (not "is newer": edits can
  carry an older timestamp than the last one seen);
- due a scheduled refresh - cached detail older than `FULL_REFRESH_DAYS` (default 7), which keeps
  the detail-only usage fields (`Active Users`, `Total sessions`, `Last Activity Date`) current.

Detail calls run on `DETAIL_WORKERS` threads (default 8) and retry 429 / 500 / 502 / 503 / 504
and network timeouts / connection errors with `Retry-After` or exponential backoff; the token
refresh is serialised under a lock. A fresh detail payload wins over the list item; for cached
detail, today's list fields win over the cache, so name, status and `isBlocked` are always
current. `Detail As Of` on every row records when its detail was fetched. Set
`DETAIL_MODE = 'full'` to re-fetch every agent.

## Large tenants: checkpoint, resume and missing detail

A first run on a large tenant has no cache, so every agent needs a detail call - about 20,000
calls takes a while even on 8 workers. Step 3 prints a rough expected duration up front, then a
progress line every `DETAIL_PROGRESS_EVERY` calls (default 500) with the measured rate and ETA.

- **Checkpoint and resume.** Every `DETAIL_CHECKPOINT_EVERY` successful calls (default 1,000),
  and again if the step fails or is interrupted, the fetched detail is MERGEd into
  `dbo.agents_365_detail_cache`. If the run stops part-way, **just rerun the notebook**: the
  checkpointed agents are reused and only the rest are fetched. Only genuinely fetched detail is
  checkpointed, each against the `lastModifiedDateTime` it was fetched for, so a failed run never
  marks a change as seen. The final cache write (after the snapshot and history) prunes agents
  that left the catalog and records creators. A `DETAIL_MODE = 'full'` run does not read the
  cache, so switch back to `'incremental'` to resume one.
- **Missing detail.** An agent whose detail call fails keeps its cached detail if it has one
  (`Detail status` = `cached - refetch failed`). If it has none, it is written **list-only**:
  its list fields are kept, usage, Bot Id and capability columns are blank, and `Detail status`
  records why, for example `missing (HTTP 424)` (Graph's *Failed Dependency*, returned for some
  packages), `missing (HTTP 404)`, `missing (HTTP 403)` or `missing (ReadTimeout)`. Step 3
  prints a warning with the count by reason. These agents are never cached, so the next run
  tries them again. 424 and 404 are not retried within a run; 429, 5xx and network errors are.
  However many agents are missing, the run carries on - with one exception: if **every** detail
  call failed with 401 or 403 and there is no cached detail at all, sign-in or consent is broken,
  so the run fails rather than write a snapshot with no detail.
- **`Detail status`** (`fetched` / `cached` / `cached - refetch failed` / `missing (<reason>)`) is written
  to `dbo.agents_365` only, not to the history table. List-only rows are kept out of the history
  merge so they never blank an agent's usage fields there.
- **Step guards.** If a step fails, the later cells stop with a message pointing back to it,
  instead of a `NameError`.

## Creator attribution (`Agent creator UPN`)

`Publisher` is a **display string**, not an identifier - it is length-limited and is not a
usable join key, even though both `Publisher` and `Agent creator` are mapped from it. To
report agents by franchise/function you need a real identity, so this notebook resolves one
and writes two extra columns:

| Column | Contents |
|---|---|
| `Agent creator UPN` | normalised (lowercased, trimmed) UPN - joins directly to `PersonId_Normalized` in `copilot_org_data` |
| `Agent creator source` | which tier resolved it, or `unattributed` |

Three tiers run in order; each only processes agents still unresolved:

| Tier | Method | Permission |
|---|---|---|
| 1 | `ownerId` -> `/users/{id}` via `$batch` | `User.Read.All` **(new)** |
| 2 | `appId` / `agentIdentityId` -> `/servicePrincipals/{id}/owners` | `Application.Read.All` (already required) |
| 3 | earliest Purview audit-log creation event | existing parsed table |

Each tier is independently switchable via the `RESOLVE_VIA_*` flags. The outcome is cached per
agent, so the tiers run only for agents whose detail was fetched this run.

**Always LEFT-join on `Agent creator UPN`.** Agents owned by a service principal or an admin
account resolve to a UPN with no row in the org table, and an inner join would silently drop
them and understate your totals. Surface `Agent creator source` as a slicer so attribution
coverage is visible on the page rather than assumed - a high unattributed rate is an
agent-ownership governance finding, not a bug.

## Owner account (governance)

`Owner account` says whether the person behind each agent can still look after it:

| Value | Meaning |
|---|---|
| `Active` | the owner's account is enabled |
| `Disabled` | the account exists but is disabled - typically someone who has left |
| `Not found` | the account no longer exists in Entra ID |
| *(blank)* | no owner on record (`unattributed`), or Graph could not answer this run |

It checks the owner's object id when tier 1 resolved it, else the resolved creator UPN, with
`/users/{id}?$select=id,accountEnabled` via `$batch` - the `User.Read.All` permission creator
resolution already uses, so no new consent. It runs on **every** run, not from the cache, because
an owner leaving is exactly the change to catch. It is written to `dbo.agents_365` only, not to
the history table. Set `CHECK_OWNER_ACCOUNT = False` to skip it.

## Optional: full field passthrough

`INCLUDE_RAW_PASSTHROUGH` (default **`False`**) additionally carries **every** field the API
returns, under its native camelCase name, with nested objects serialised to JSON. Fields are
unioned across all agents, so an attribute present on only some agents still becomes a column.

It is **off by default on purpose.** The dashboard's `Agents 365` query is additive - it has
no `Table.SelectColumns` - so every column in `dbo.agents_365` reaches the semantic model.
Turning passthrough on adds roughly 30 unmodelled columns to that table. Enable it when you
need an attribute the canonical schema does not carry, and expect the model to widen.

The canonical columns are never clobbered: where a raw key collides with one, the canonical
value wins, and a raw value that genuinely differs is kept under `<key>_raw`.

## 1. Configuration & app-only sign-in

**App-only (client-credentials)** flow - runs unattended, so this notebook can be a scheduled Fabric
job. No device-code, no browser. The service principal's admin-consented **Application** permissions
(`CopilotPackages.Read.All` + `Application.Read.All`) are carried in the token via the `.default`
scope.

````python
# === CONFIG ===
TENANT_ID    = '<your-tenant-guid>'
CLIENT_ID    = '<app-reg-client-id>'   # App perms (admin-consented): CopilotPackages.Read.All
                                       #   + Application.Read.All + User.Read.All
TARGET_TABLE = 'dbo.agents_365'
WRITE_MODE   = 'overwrite'
ALLOW_EMPTY_SNAPSHOT = False

# Incremental detail pull. The list call is cheap and runs in full every time; the
# per-agent detail call is the slow part, so it only runs for agents that are new,
# have no cache entry, or whose lastModifiedDateTime differs from the cached value.
# Usage fields (Active Users, Total sessions, Last Activity Date) only come from the
# detail call, so every agent is re-fetched at least every FULL_REFRESH_DAYS days.
# 'Detail As Of' on each row shows how fresh its detail fields are.
DETAIL_MODE        = 'incremental'   # 'incremental' | 'full' (re-fetch every agent)
FULL_REFRESH_DAYS  = 7
DETAIL_WORKERS     = 8               # parallel detail calls
DETAIL_CACHE_TABLE = 'dbo.agents_365_detail_cache'
HISTORY_TABLE      = 'dbo.agents_365_history'   # '' disables the history merge

# Large tenants (tens of thousands of agents): successful detail calls are saved to
# DETAIL_CACHE_TABLE every DETAIL_CHECKPOINT_EVERY calls, so a run that fails or is
# interrupted resumes on rerun instead of starting again. Progress prints every
# DETAIL_PROGRESS_EVERY calls.
DETAIL_CHECKPOINT_EVERY = 1000
DETAIL_PROGRESS_EVERY   = 500
# An agent whose detail call fails (424 Failed Dependency, 404, 403, or a 5xx / timeout
# after retries) and that has no cached detail is written list-only with
# 'Detail status' = 'missing (<reason>)' and retried next run; it does not fail the run.

# Graph throttling and transient faults: 429/500/502/503/504 and network timeouts /
# connection errors are retried with Retry-After (or exponential backoff).
RETRY_STATUSES = (429, 500, 502, 503, 504)
MAX_RETRIES    = 6

# Creator resolution toggles - each tier is independent and skippable.
RESOLVE_VIA_OWNER_ID   = True    # tier 1: ownerId -> /users
RESOLVE_VIA_SP_OWNERS  = True    # tier 2: appId -> /servicePrincipals/{id}/owners
RESOLVE_VIA_AUDIT_LOG  = True    # tier 3: Purview creation events
AUDIT_TABLE            = 'dbo.copilot_interactions_parsed'

# Governance: is each agent's owner still an active account? Checked on every run
# (an owner can leave between detail refreshes) and written to 'Owner account'.
# Uses User.Read.All, which creator resolution already needs.
CHECK_OWNER_ACCOUNT    = True

# Optional: carry every raw API field through alongside the canonical schema.
# Off by default. The dashboard's `Agents 365` query is additive (no Table.SelectColumns),
# so every column here reaches the semantic model -- enabling this widens it by ~30 columns.
# Turn it on when you need an attribute the canonical schema does not carry.
INCLUDE_RAW_PASSTHROUGH = False

# Client secret - DO NOT hardcode. Pull from Key Vault at runtime, e.g.:
#   CLIENT_SECRET = notebookutils.credentials.getSecret('https://<vault>.vault.azure.net/', 'Agent365AppSecret')
CLIENT_SECRET = '<from-key-vault>'

import random
import threading
import time
import requests

_TOKEN_CACHE = {'value': '', 'expires_at': 0.0}
# Detail calls run on DETAIL_WORKERS threads; one lock means one mint when the token
# expires, instead of every worker racing to the token endpoint at once.
_TOKEN_LOCK = threading.Lock()


def _get_graph_token(force_refresh: bool = False, stale_token: str = None) -> str:
    """Return an app-only Graph token, re-minting it when the cached one is stale.

    Client-credentials tokens are short-lived (commonly ~60 minutes) and a full
    registry snapshot issues one request per package, so a large tenant can easily
    run past the expiry of a token minted once at the top of the notebook - every
    later call then fails with 401. Caching against the real 'expires_in' keeps
    long runs alive without re-authenticating on every request.

    `stale_token` is the token a caller saw rejected: if another thread has already
    replaced it, the forced refresh reuses that newer token rather than minting again.
    """
    with _TOKEN_LOCK:
        now = time.time()
        cached = _TOKEN_CACHE['value']
        live = bool(cached) and now < _TOKEN_CACHE['expires_at']
        if live and (not force_refresh or (stale_token is not None and cached != stale_token)):
            return cached
        url  = f'https://login.microsoftonline.com/{TENANT_ID}/oauth2/v2.0/token'
        data = {
            'client_id':     CLIENT_ID,
            'client_secret': CLIENT_SECRET,
            'scope':         'https://graph.microsoft.com/.default',
            'grant_type':    'client_credentials',
        }
        r = requests.post(url, data=data, timeout=60)
        r.raise_for_status()
        payload = r.json()
        token = payload['access_token']
        try:
            expires_in = float(payload.get('expires_in', 3600))
        except (TypeError, ValueError):
            expires_in = 3600.0
        # Renew early so an in-flight request cannot straddle the expiry boundary.
        _TOKEN_CACHE['value'] = token
        _TOKEN_CACHE['expires_at'] = now + max(expires_in - 300.0, 60.0)
        return token


def graph_headers(force_refresh: bool = False, stale_token: str = None) -> dict:
    return {'Authorization': f'Bearer {_get_graph_token(force_refresh, stale_token)}'}


def _retry_delay(response, attempt):
    """Seconds to wait before retrying a throttled call: Retry-After, else backoff."""
    headers = getattr(response, 'headers', None) or {}
    retry_after = headers.get('Retry-After') if hasattr(headers, 'get') else None
    if retry_after is not None:
        try:
            return min(max(float(retry_after), 0.0), 120.0)
        except (TypeError, ValueError):
            pass
    return min(2.0 ** attempt, 60.0) + random.uniform(0.0, 1.0)


def graph_request(method: str, url: str, **kwargs):
    """Call Graph with a live token.

    - 401: retried once with a freshly minted token. A 401 is ambiguous - an expired
      token (recoverable) or absent consent (fatal) - so the caller judges the result.
    - RETRY_STATUSES (429 / 500 / 502 / 503 / 504): throttling or a transient server
      fault, retried up to MAX_RETRIES times, honouring Retry-After.
    - requests Timeout / ConnectionError: retried on the same backoff, then re-raised.
    Other statuses return immediately, including a 404 or 424 Failed Dependency on one
    package: those are not transient, so step 3 writes that agent list-only instead.
    """
    kwargs.setdefault('timeout', 60)
    extra = kwargs.pop('headers', None) or {}
    headers = graph_headers()
    reauthenticated = False
    attempt = 0
    while True:
        try:
            response = requests.request(method, url, headers={**headers, **extra}, **kwargs)
        except (requests.Timeout, requests.ConnectionError):
            if attempt >= MAX_RETRIES:
                raise
            time.sleep(_retry_delay(None, attempt))
            attempt += 1
            headers = graph_headers()
            continue
        if response.status_code == 401 and not reauthenticated:
            reauthenticated = True
            rejected = headers['Authorization'].split(' ', 1)[-1]
            headers = graph_headers(True, rejected)
            continue
        if response.status_code in RETRY_STATUSES and attempt < MAX_RETRIES:
            time.sleep(_retry_delay(response, attempt))
            attempt += 1
            headers = graph_headers()
            continue
        return response


# Fail fast on bad credentials. Deliberately keeps no token string: every Graph
# call goes through graph_request(), which re-mints on expiry - a module-level
# token would go stale mid-run and invite exactly the 401s this cell guards against.
_startup_token_ok = bool(_get_graph_token())
print('app-only token acquired:', _startup_token_ok)
````

## 2. Call the v1.0 catalog endpoint

Unchanged from upstream. Uses `/v1.0` (GA, app-only), auto-falls back to `/beta`.

- **403** = missing Agent 365 licence (SKU check), or no admin consent for `CopilotPackages.Read.All`.
- **401** = token/consent problem, not a code bug.

````python
import requests

# Step guards: each step clears its own flag and every later one, and sets its own only
# when it finishes - so a failed step stops the cells after it with a clear message.
if 'graph_request' not in globals():
    raise RuntimeError('Step 1 (configuration & sign-in) has not run in this session - run it first.')
_STEP2_OK = _STEP3_OK = _STEP4_OK = False

API_VERSION = 'v1.0'


def _base(v):
    return f'https://graph.microsoft.com/{v}/copilot/admin/catalog/packages'


def _validate_catalog_page(data, page_number):
    if not isinstance(data, dict):
        raise ValueError(f'Agent 365 catalog page {page_number} did not return an object.')
    if 'value' not in data:
        raise ValueError(f"Agent 365 catalog page {page_number} is missing required 'value'.")
    value = data['value']
    if not isinstance(value, list):
        raise ValueError(f"Agent 365 catalog page {page_number} returned a non-list 'value'.")
    next_link = data.get('@odata.nextLink')
    if next_link is not None and (not isinstance(next_link, str) or not next_link.strip()):
        raise ValueError(f"Agent 365 catalog page {page_number} returned an invalid '@odata.nextLink'.")
    for item_number, item in enumerate(value, start=1):
        if not isinstance(item, dict):
            raise ValueError(f'Agent 365 catalog page {page_number} item {item_number} is not an object.')
    return value, next_link


BASE = _base(API_VERSION)
# Every Graph call goes through graph_request so the token stays live.

probe = graph_request('GET', f'{BASE}?$top=1')
if probe.status_code == 404 and API_VERSION == 'v1.0':
    API_VERSION = 'beta'
    BASE = _base(API_VERSION)
    probe = graph_request('GET', f'{BASE}?$top=1')
if probe.status_code == 403:
    raise PermissionError('403: Agent 365 catalog not accessible. Requires an Agent 365 LICENCE in '
                          'the tenant (SKU check) AND admin-consented Application permission '
                          'CopilotPackages.Read.All.')
if probe.status_code == 401:
    raise PermissionError('401 after a token refresh, so this is not expiry: check admin '
                          'consent for CopilotPackages.Read.All + Application.Read.All '
                          '+ User.Read.All.')
probe.raise_for_status()
print(f'Using {API_VERSION} endpoint.')

packages, url, seen_urls = [], BASE, set()
for page_number in range(1, 501):
    if url in seen_urls:
        raise ValueError(f'Agent 365 catalog paging loop detected at page {page_number}.')
    seen_urls.add(url)
    r = graph_request('GET', url)
    r.raise_for_status()
    page, next_url = _validate_catalog_page(r.json(), page_number)
    packages.extend(page)
    url = next_url
    if not url:
        break
else:
    raise ValueError('Agent 365 catalog exceeded the 500-page safety cap.')

print('packages in catalog:', len(packages))
_STEP2_OK = True
````

## 3. Fetch per-package detail (incremental, parallel) and flatten every field

The list endpoint returns inventory metadata; the detail endpoint adds usage metrics
and `elementDetails`. Detail is fetched only for new, changed, uncached or refresh-due
agents, on `DETAIL_WORKERS` threads; everyone else reuses the cached detail with today's
list fields on top. A fresh payload wins over the list item. The union of fields is
flattened, so nothing the API returns is discarded.

**Large tenants:** progress prints every `DETAIL_PROGRESS_EVERY` calls with an ETA, and
successful calls are checkpointed to the detail cache every `DETAIL_CHECKPOINT_EVERY`, so if
this step fails or is interrupted, rerunning resumes rather than starting again. Agents with no
detail and no cached copy (a 424, 404, 403, or a 5xx / timeout after retries) are written
list-only with `Detail status` = `missing (<reason>)` and retried next run, without failing the
step. Only an all-401/403 run with no cache at all fails here.

````python
if not globals().get('_STEP2_OK'):
    raise RuntimeError('Step 2 (catalog list) did not complete in this session, so there is no '
                       '`packages` list - fix the error shown under step 2 and rerun it first.')
_STEP3_OK = _STEP4_OK = False

import json as _json
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime, timedelta, timezone
from time import monotonic as _monotonic


def _stringify(value):
    # Scalars -> str. Nested structures -> compact JSON so nothing is lost.
    if value is None:
        return ''
    if isinstance(value, bool):
        return 'true' if value else 'false'
    if isinstance(value, (dict, list)):
        return _json.dumps(value, ensure_ascii=False, sort_keys=True)
    return str(value)


def _flatten_raw(payload):
    # One flat dict of every top-level API field, native camelCase names preserved.
    flat = {}
    for key, value in payload.items():
        if key.startswith('@odata'):
            continue
        flat[key] = _stringify(value)
    return flat


def _package_id(package):
    package_id = package.get('id') or package.get('titleId') or package.get('packageId')
    if package_id in (None, ''):
        raise ValueError('Agent 365 catalog list item is missing id/titleId/packageId.')
    return str(package_id)


def _list_modified(package):
    return str(package.get('lastModifiedDateTime') or '').strip()


def _parse_utc(value):
    try:
        parsed = datetime.fromisoformat(str(value).strip().replace('Z', '+00:00'))
    except (TypeError, ValueError):
        return None
    return parsed if parsed.tzinfo else parsed.replace(tzinfo=timezone.utc)


def _iso_utc(moment):
    return moment.astimezone(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ')


def _select_detail_targets(list_items, cache, now, mode='incremental', full_refresh_days=7):
    """Decide which agents need a detail call, and why.

    Returns {package_id: reason}. An agent is fetched when it is new / uncached, when
    its lastModifiedDateTime *differs* from the cached value (not 'is newer': edits
    can carry an older timestamp than the last one seen), or when its cached detail
    is older than full_refresh_days - which keeps the detail-only usage fields fresh.
    """
    if mode not in ('incremental', 'full'):
        raise ValueError(f"DETAIL_MODE must be 'incremental' or 'full', not {mode!r}.")
    cutoff = now - timedelta(days=full_refresh_days)
    targets = {}
    for package in list_items:
        package_id = _package_id(package)
        if package_id in targets:
            continue
        entry = cache.get(package_id)
        if mode == 'full':
            targets[package_id] = 'full mode'
        elif entry is None:
            targets[package_id] = 'new / uncached'
        elif _list_modified(package) != entry.get('lastModifiedDateTime', ''):
            targets[package_id] = 'changed'
        else:
            as_of = _parse_utc(entry.get('detailAsOf'))
            if as_of is None or as_of <= cutoff:
                targets[package_id] = 'scheduled refresh'
    return targets


def _merge_fresh(package, detail):
    """Fresh fetch: the detail payload wins over the list item (it is the richer one)."""
    merged = dict(package)
    merged.update(detail)
    return merged


def _merge_cached(package, cached_detail):
    """Cached detail: today's list fields win, so name/status/blocked stay current."""
    merged = dict(cached_detail)
    merged.update(package)
    return merged


def _fresh_cache_entry(package, detail, now_iso):
    """Cache entry for detail genuinely fetched this run, keyed to the list's
    lastModifiedDateTime it was fetched against. Creator fields start blank, so the
    creator tiers run for the agent (step 4 fills them in before the final cache write)."""
    return {
        'detail': detail,
        'lastModifiedDateTime': _list_modified(package),
        'detailAsOf': now_iso,
        'creatorUpn': '',
        'creatorSource': '',
    }


def _fmt_duration(seconds):
    seconds = max(int(round(seconds)), 0)
    hours, rest = divmod(seconds, 3600)
    minutes, secs = divmod(rest, 60)
    if hours:
        return f'{hours}h {minutes:02d}m'
    if minutes:
        return f'{minutes}m {secs:02d}s'
    return f'{secs}s'


def _progress_line(done, total, failures, elapsed):
    rate = done / elapsed if elapsed > 0 else 0.0
    eta = _fmt_duration((total - done) / rate) if rate > 0 else 'unknown'
    return (f'  detail calls {done}/{total} ({done / total * 100:.0f}%) | {rate:.1f}/s | '
            f'failed {failures} | elapsed {_fmt_duration(elapsed)} | ETA {eta}')


def _http_status(exc):
    return getattr(getattr(exc, 'response', None), 'status_code', None)


def _failure_reason(exc):
    status = _http_status(exc)
    return f'HTTP {status}' if status is not None else type(exc).__name__


def _failure_census(failed):
    """{reason: count} for {package_id: exception}, most common first."""
    census = {}
    for exc in failed.values():
        reason = _failure_reason(exc)
        census[reason] = census.get(reason, 0) + 1
    return dict(sorted(census.items(), key=lambda item: (-item[1], item[0])))


def _fetch_details(package_ids, fetch, workers, on_batch=None, batch_size=0,
                   progress=None, progress_every=0, clock=None):
    """Run `fetch(package_id)` on a thread pool. Returns ({id: detail}, {id: error}).

    on_batch({id: detail}) is called on this thread with every `batch_size` new
    successes, and once more with any remainder when the pool finishes, fails or is
    interrupted - so whatever was fetched is saved even if the run dies later. A
    failing on_batch only warns; its entries are retried with the next batch.
    progress(done, total, failures, elapsed_seconds) runs every `progress_every` calls.
    """
    fetched, failed, pending = {}, {}, {}
    if not package_ids:
        return fetched, failed
    clock = clock or _monotonic
    started, total = clock(), len(package_ids)
    flush_at = [batch_size]

    def flush():
        if not (on_batch and pending):
            return
        batch = dict(pending)
        try:
            on_batch(batch)
        except Exception as exc:          # a checkpoint failure must not stop the fetch
            flush_at[0] = len(pending) + batch_size
            print(f'[warn] detail checkpoint of {len(batch)} agent(s) failed ({exc}); '
                  'retrying with the next batch.')
            return
        for package_id in batch:
            pending.pop(package_id, None)
        flush_at[0] = batch_size

    pool = ThreadPoolExecutor(max_workers=max(1, int(workers)))
    try:
        futures = {pool.submit(fetch, package_id): package_id for package_id in package_ids}
        for done, future in enumerate(as_completed(futures), 1):
            package_id = futures[future]
            try:
                fetched[package_id] = pending[package_id] = future.result()
            except Exception as exc:      # collected and judged below, never swallowed
                failed[package_id] = exc
            if batch_size and len(pending) >= flush_at[0]:
                flush()
            if progress and progress_every and (done % progress_every == 0 or done == total):
                progress(done, total, len(failed), clock() - started)
    finally:
        pool.shutdown(wait=False, cancel_futures=True)
        flush()
    return fetched, failed


def _assemble_details(list_items, cache, fetched, failed, now_iso):
    """Combine today's list with fresh or cached detail.

    Returns (details, meta, new_cache). `meta[i]` carries detailAsOf, whether row i
    was fetched this run, and detailStatus: fetched | cached | cached - refetch failed
    | missing (<reason>). `new_cache` holds only agents still in today's list, so agents
    removed from the catalog drop out of the cache. A failed detail call falls back to
    the cached copy, keeping the cached lastModifiedDateTime so the change is retried.

    An agent with neither fresh nor cached detail - its call failed with a 424, 404,
    403, or a 5xx / timeout after retries - is written list-only, with detailStatus
    'missing (HTTP 424)' (or the exception type) and a blank detailAsOf, and is kept
    out of the cache so the next run retries it. That never fails the run, with one
    exception: when every row would be list-only and every failure was a 401 or 403,
    sign-in or consent is broken, so the run fails rather than write a snapshot with
    no detail at all.
    """
    details, meta, new_cache = [], [], {}
    for package in list_items:
        package_id = _package_id(package)
        entry = cache.get(package_id)
        if package_id in fetched:
            detail = fetched[package_id]
            details.append(_merge_fresh(package, detail))
            meta.append({'packageId': package_id, 'detailAsOf': now_iso, 'fetched': True,
                         'detailStatus': 'fetched'})
            new_cache[package_id] = _fresh_cache_entry(package, detail, now_iso)
            continue
        if entry is None:
            reason = _failure_reason(failed[package_id]) if package_id in failed else 'no detail'
            details.append(dict(package))
            meta.append({'packageId': package_id, 'detailAsOf': '', 'fetched': False,
                         'detailStatus': f'missing ({reason})'})
            continue
        refetch_failed = package_id in failed
        details.append(_merge_cached(package, entry['detail']))
        meta.append({'packageId': package_id, 'detailAsOf': entry.get('detailAsOf', ''),
                     'fetched': False, 'refetchFailed': refetch_failed,
                     'detailStatus': 'cached - refetch failed' if refetch_failed else 'cached'})
        # Keep the cached lastModifiedDateTime when a refetch failed, so the change is
        # retried on the next run instead of being silently marked as seen.
        new_cache[package_id] = dict(entry)
    if (failed and all(m['detailStatus'].startswith('missing') for m in meta)
            and all(_http_status(exc) in (401, 403) for exc in failed.values())):
        raise RuntimeError(
            f'Every Agent 365 detail call ({len(failed)}) failed with HTTP 401/403 '
            f'({_failure_census(failed)}) and there is no cached detail to fall back on, so no '
            'snapshot is written: it would carry no detail at all. This is a sign-in or consent '
            'problem, not a per-agent one - check that CopilotPackages.Read.All is granted and '
            'admin-consented, then rerun.')
    return details, meta, new_cache


def _cache_from_rows(rows):
    """Rebuild the cache dict from agents_365_detail_cache rows; bad rows are skipped."""
    cache = {}
    for row in rows:
        try:
            detail = _json.loads(row['detail_json'])
        except (TypeError, ValueError, KeyError):
            continue
        if not isinstance(detail, dict) or not row.get('package_id'):
            continue
        cache[str(row['package_id'])] = {
            'detail': detail,
            'lastModifiedDateTime': row.get('last_modified') or '',
            'detailAsOf': row.get('detail_as_of') or '',
            'creatorUpn': row.get('creator_upn') or '',
            'creatorSource': row.get('creator_source') or '',
        }
    return cache


def _cache_to_rows(cache):
    return [
        {
            'package_id': package_id,
            'last_modified': entry.get('lastModifiedDateTime', ''),
            'detail_as_of': entry.get('detailAsOf', ''),
            'creator_upn': entry.get('creatorUpn', ''),
            'creator_source': entry.get('creatorSource', ''),
            'detail_json': _json.dumps(entry['detail'], ensure_ascii=False, sort_keys=True),
        }
        for package_id, entry in sorted(cache.items())
    ]


def _load_detail_cache(table_name):
    if DETAIL_MODE == 'full' or not table_name:
        return {}
    try:
        if not spark.catalog.tableExists(table_name):
            return {}
        return _cache_from_rows([r.asDict() for r in spark.table(table_name).collect()])
    except Exception as exc:
        print(f'[warn] detail cache {table_name} unreadable ({exc}); fetching every agent.')
        return {}


_DETAIL_CACHE_SCHEMA = ('package_id string, last_modified string, detail_as_of string, '
                        'creator_upn string, creator_source string, detail_json string')


def _checkpoint_detail_cache(entries, table_name):
    """Upsert freshly fetched detail into the cache table mid-run (MERGE on package_id).

    Only genuinely fetched entries reach here, each keyed to the lastModifiedDateTime
    it was fetched against, so a run that fails later can never mark a change as seen
    with stale detail. The final write in step 5 still prunes removed agents and
    records creators.
    """
    if not table_name or not entries:
        return
    batch_df = spark.createDataFrame(_cache_to_rows(entries), _DETAIL_CACHE_SCHEMA)
    if not spark.catalog.tableExists(table_name):
        batch_df.write.mode('overwrite').format('delta').saveAsTable(table_name)
        return
    from delta.tables import DeltaTable
    (DeltaTable.forName(spark, table_name).alias('t')
        .merge(batch_df.alias('s'), 't.package_id = s.package_id')
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute())


def _fetch_one_detail(package_id):
    response = graph_request('GET', f'{BASE}/{package_id}')
    response.raise_for_status()
    detail = response.json()
    if not isinstance(detail, dict):
        raise ValueError(f'Agent 365 detail for {package_id!r} did not return an object.')
    return detail


# Settings added for large tenants; defaults apply if an older config cell is kept.
_CHECKPOINT_EVERY = globals().get('DETAIL_CHECKPOINT_EVERY', 1000)
_PROGRESS_EVERY   = globals().get('DETAIL_PROGRESS_EVERY', 500)
_EST_SECONDS_PER_CALL = 0.5   # rough Graph latency, for the up-front estimate only

run_started = datetime.now(timezone.utc)
RUN_AS_OF = _iso_utc(run_started)

detail_cache = _load_detail_cache(DETAIL_CACHE_TABLE)
targets = _select_detail_targets(packages, detail_cache, run_started, DETAIL_MODE, FULL_REFRESH_DAYS)
reasons = {}
for _reason in targets.values():
    reasons[_reason] = reasons.get(_reason, 0) + 1
print(f'cached agents: {len(detail_cache)} | detail calls needed: {len(targets)} of '
      f'{len(packages)} ' + (str(reasons) if reasons else ''))

_packages_by_id = {}
for _package in packages:
    _packages_by_id.setdefault(_package_id(_package), _package)

if targets:
    _workers = max(1, int(DETAIL_WORKERS))
    print(f'expected duration: roughly {_fmt_duration(len(targets) * _EST_SECONDS_PER_CALL / _workers)} '
          f'({len(targets)} calls on {_workers} workers at ~{_EST_SECONDS_PER_CALL}s each; the '
          'progress lines below show the measured rate and ETA)')
    if DETAIL_CACHE_TABLE:
        print(f'successful calls are checkpointed to {DETAIL_CACHE_TABLE} every {_CHECKPOINT_EVERY}: '
              'if this run fails or is interrupted, rerun it and only what is missing is fetched'
              + (" (set DETAIL_MODE = 'incremental' first)" if DETAIL_MODE == 'full' else ''))

_checkpointed = [0]


def _checkpoint(batch):
    entries = {pid: _fresh_cache_entry(_packages_by_id[pid], detail, RUN_AS_OF)
               for pid, detail in batch.items()}
    _checkpoint_detail_cache(entries, DETAIL_CACHE_TABLE)
    _checkpointed[0] += len(entries)


fetched_details, failed_details = _fetch_details(
    list(targets), _fetch_one_detail, DETAIL_WORKERS,
    on_batch=_checkpoint if DETAIL_CACHE_TABLE else None, batch_size=_CHECKPOINT_EVERY,
    progress=lambda *args: print(_progress_line(*args)), progress_every=_PROGRESS_EVERY)
if _checkpointed[0]:
    print(f'checkpointed {_checkpointed[0]} fetched agent(s) to {DETAIL_CACHE_TABLE}')
if failed_details:
    print(f'[warn] {len(failed_details)} detail call(s) returned no detail (429, 5xx and network '
          f'errors after retries; 404 and 424 are not retried): {_failure_census(failed_details)}')

details, detail_meta, new_detail_cache = _assemble_details(
    packages, detail_cache, fetched_details, failed_details, RUN_AS_OF)

_status_counts = {}
for _info in detail_meta:
    _status_counts[_info['detailStatus']] = _status_counts.get(_info['detailStatus'], 0) + 1
print(f'detail status: {_status_counts}')
_missing_ids = [m['packageId'] for m in detail_meta if m['detailStatus'].startswith('missing')]
if _missing_ids:
    _by_reason = {}
    for _info in detail_meta:
        if _info['detailStatus'].startswith('missing'):
            _by_reason[_info['detailStatus']] = _by_reason.get(_info['detailStatus'], 0) + 1
    _by_reason = dict(sorted(_by_reason.items(), key=lambda item: (-item[1], item[0])))
    print(f"[warn] {len(_missing_ids)} agent(s) have no detail and no cached copy: written list-only "
          "(usage, Bot Id and capability columns blank) and retried next run. "
          f"By 'Detail status': {_by_reason}. First: "
          + ', '.join(_missing_ids[:10]) + (' ...' if len(_missing_ids) > 10 else ''))
if _status_counts.get('cached - refetch failed'):
    print(f"[warn] {_status_counts['cached - refetch failed']} agent(s) kept their cached detail "
          'after a failed refetch; retried next run.')

_all_api_fields = sorted({k for d in details for k in d if not k.startswith('@odata')})
print(f'distinct API fields observed: {len(_all_api_fields)}')
print('  ' + ', '.join(_all_api_fields))
_STEP3_OK = True
````

## 4. Resolve the agent creator to a joinable UPN

`Publisher` is a truncated display string. This produces `Agent creator UPN`
(lowercased + trimmed, matching the repo's `_normalise_identity`) so it joins
straight to `PersonId_Normalized` in the org table.

Tiers run in order and each only processes agents still unresolved.

````python
def _normalise_identity(value):
    return (value or '').strip().lower()


# --- Tier 1: ownerId -> /users via $batch -------------------------------------
def _resolve_owner_ids(owner_ids):
    resolved = {}
    owner_ids = [o for o in owner_ids if o]
    for i in range(0, len(owner_ids), 15):
        chunk = owner_ids[i:i + 15]
        batch = {'requests': [
            {'id': str(n), 'method': 'GET',
             'url': f'/users/{oid}?$select=id,userPrincipalName'}
            for n, oid in enumerate(chunk)
        ]}
        resp = graph_request('POST', 'https://graph.microsoft.com/v1.0/$batch',
                             headers={'Content-Type': 'application/json'},
                             json=batch)
        if resp.status_code != 200:
            print(f'  tier1: batch failed ({resp.status_code}) - skipping chunk')
            continue
        for item in resp.json().get('responses', []):
            if item.get('status') != 200:
                continue
            body = item.get('body') or {}
            upn = _normalise_identity(body.get('userPrincipalName'))
            if body.get('id') and upn:
                resolved[body['id']] = upn
    return resolved


# --- Owner account: is the owner still an active account? -------------------
def _owner_key(owner_id, creator_upn, creator_source):
    """The identity to check: the owner's object id when tier 1 resolved it (stable
    across renames), else the resolved creator UPN. Unattributed agents have none."""
    if creator_source == 'ownerId' and owner_id:
        return owner_id
    return creator_upn or ''


def _owner_account_status(keys):
    """Map each owner key (object id or UPN) to 'Active', 'Disabled' or 'Not found'.

    A key Graph could not answer for - a failed batch, a throttled or other error
    response - is left out, so it reads as unknown rather than as an owner who left.
    """
    from urllib.parse import quote
    status = {}
    keys = sorted({k for k in keys if k})
    for i in range(0, len(keys), 20):
        chunk = keys[i:i + 20]
        batch = {'requests': [
            {'id': str(n), 'method': 'GET',
             'url': f"/users/{quote(key, safe='@')}?$select=id,accountEnabled"}
            for n, key in enumerate(chunk)
        ]}
        resp = graph_request('POST', 'https://graph.microsoft.com/v1.0/$batch',
                             headers={'Content-Type': 'application/json'},
                             json=batch)
        if resp.status_code != 200:
            print(f'  owner account: batch failed ({resp.status_code}) - left unknown')
            continue
        for item in resp.json().get('responses', []):
            try:
                key = chunk[int(item.get('id'))]
            except (TypeError, ValueError, IndexError):
                continue
            if item.get('status') == 404:
                status[key] = 'Not found'
            elif item.get('status') == 200:
                enabled = (item.get('body') or {}).get('accountEnabled')
                if enabled is True:
                    status[key] = 'Active'
                elif enabled is False:
                    status[key] = 'Disabled'
    return status


# --- Tier 2: appId / agentIdentityId -> servicePrincipal owners ---------------
_sp_cache = {}

def _resolve_sp_owner(app_id, agent_identity_id):
    cache_key = (app_id or '', agent_identity_id or '')
    if cache_key in _sp_cache:
        return _sp_cache[cache_key]
    result = ''
    for ident, by_app in ((app_id, True), (agent_identity_id, False)):
        if not ident or result:
            continue
        url = (f"https://graph.microsoft.com/v1.0/servicePrincipals(appId='{ident}')"
               if by_app else
               f'https://graph.microsoft.com/v1.0/servicePrincipals/{ident}')
        sp = graph_request('GET', url)
        if sp.status_code != 200:
            continue
        sp_id = (sp.json() or {}).get('id')
        if not sp_id:
            continue
        owners = graph_request(
            'GET',
            f'https://graph.microsoft.com/v1.0/servicePrincipals/{sp_id}/owners'
            '?$select=id,userPrincipalName')
        if owners.status_code != 200:
            continue
        for owner in owners.json().get('value', []):
            upn = _normalise_identity(owner.get('userPrincipalName'))
            if upn:
                result = upn
                break
    _sp_cache[cache_key] = result
    return result


# --- Tier 3: Purview audit-log earliest creation event ------------------------
def _audit_creator_map(keys_wanted):
    # Earliest audit event per agent identifier -> creator UPN. Defensive: the
    # parsed table's column names vary by ingester version, so detect rather than
    # assume, and return empty if the shape is not recognisable.
    if not keys_wanted:
        return {}
    try:
        if not spark.catalog.tableExists(AUDIT_TABLE):
            print(f'  tier3: {AUDIT_TABLE} not present - skipping')
            return {}
    except Exception as exc:
        print(f'  tier3: cannot inspect {AUDIT_TABLE} ({exc}) - skipping')
        return {}

    from pyspark.sql import functions as F

    audit = spark.table(AUDIT_TABLE)
    cols = {c.lower(): c for c in audit.columns}

    user_col = next((cols[c] for c in (
        'userid', 'userprincipalname', 'upn', 'userkey', 'createdby',
        'audit_userid_normalized', 'audit_userid') if c in cols), None)
    date_col = next((cols[c] for c in (
        'creationdate', 'createddatetime', 'interactiondate') if c in cols), None)
    agent_col = next((cols[c] for c in (
        'titleid', 'agentid', 'appid', 'copilotagentid', 'source_resourcekey',
        'agent_titleid') if c in cols), None)

    if not (user_col and date_col and agent_col):
        print('  tier3: audit table lacks user/date/agent columns - skipping')
        print(f'         available: {", ".join(sorted(audit.columns))[:300]}')
        return {}

    print(f'  tier3: using {agent_col} / {user_col} / {date_col}')
    # Order on a real timestamp. Casting to string sorts lexicographically, which only
    # coincides with chronological order for ISO-8601 input; anything else silently
    # picks the wrong "earliest" event and so the wrong creator.
    window = audit.select(
        F.lower(F.trim(F.col(agent_col).cast('string'))).alias('k'),
        F.lower(F.trim(F.col(user_col).cast('string'))).alias('upn'),
        F.col(date_col).cast('timestamp').alias('ts'),
    ).where(F.col('k').isNotNull() & (F.col('k') != '') &
            F.col('upn').isNotNull() & (F.col('upn') != '') &
            F.col('ts').isNotNull())

    if window.isEmpty():
        print(f'  tier3: no rows with a parseable {date_col}; skipping')
        return {}

    earliest = (window.groupBy('k')
                      .agg(F.min(F.struct('ts', 'upn')).alias('first'))
                      .select('k', F.col('first.upn').alias('upn')))

    wanted = {k.lower() for k in keys_wanted if k}
    return {r['k']: r['upn'] for r in earliest.collect() if r['k'] in wanted}
````

````python
# === RESOLVE CREATORS =========================================================
if not globals().get('_STEP3_OK'):
    raise RuntimeError('Step 3 (agent detail) did not complete in this session, so there are no '
                       '`details` to resolve creators for. Fix the error shown under step 3 and '
                       'rerun it - successful detail calls are checkpointed, so it resumes.')
_STEP4_OK = False

creator_upn    = {}   # index in `details` -> upn
creator_source = {}   # index in `details` -> which tier resolved it

def _detail_title_id(detail):
    raw = detail.get('id') or detail.get('titleId') or detail.get('packageId')
    if raw in (None, ''):
        raise ValueError('Agent 365 detail is missing id/titleId/packageId.')
    raw = str(raw)
    return raw if raw.startswith(('T_', 'P_')) else f'T_{raw}'


def _seed_creators_from_cache(meta, cache):
    """Reuse the cached creator for agents whose detail was not re-fetched this run.

    Returns (upn_by_index, source_by_index, settled_indexes). Settled agents skip the
    tiers - including ones cached as unattributed, which are retried at their next
    scheduled detail refresh rather than costing Graph calls on every run.
    """
    upns, sources, settled = {}, {}, set()
    for idx, info in enumerate(meta):
        if info.get('fetched'):
            continue
        entry = cache.get(info['packageId']) or {}
        source = entry.get('creatorSource') or ''
        if not source:
            continue
        settled.add(idx)
        if source != 'unattributed' and entry.get('creatorUpn'):
            upns[idx], sources[idx] = entry['creatorUpn'], source
    return upns, sources, settled


title_ids = [_detail_title_id(d) for d in details]
creator_upn, creator_source, _creator_settled = _seed_creators_from_cache(detail_meta, new_detail_cache)
print(f'creators reused from cache: {len(_creator_settled)}')

def _needs_creator(idx):
    return idx not in _creator_settled and idx not in creator_upn

# Tier 1
if RESOLVE_VIA_OWNER_ID:
    before = len(creator_upn)
    owner_ids = [d.get('ownerId') or '' for d in details]
    pending = {o for idx, o in enumerate(owner_ids) if o and _needs_creator(idx)}
    print(f'tier1: ownerId populated on {sum(1 for o in owner_ids if o)}/{len(details)} agents, '
          f'{len(pending)} to resolve')
    if pending:
        upn_map = _resolve_owner_ids(sorted(pending))
        for idx, oid in enumerate(owner_ids):
            if not _needs_creator(idx):
                continue
            upn = upn_map.get(oid, '')
            if upn:
                creator_upn[idx], creator_source[idx] = upn, 'ownerId'
    print(f'tier1: resolved {len(creator_upn) - before}')

# Tier 2
if RESOLVE_VIA_SP_OWNERS:
    before = len(creator_upn)
    for idx, detail in enumerate(details):
        if not _needs_creator(idx):
            continue
        upn = _resolve_sp_owner(detail.get('appId'), detail.get('agentIdentityId'))
        if upn:
            creator_upn[idx], creator_source[idx] = upn, 'servicePrincipalOwner'
    print(f'tier2: resolved {len(creator_upn) - before} more')

# Tier 3
if RESOLVE_VIA_AUDIT_LOG:
    before = len(creator_upn)
    unresolved_keys = set()
    for idx, detail in enumerate(details):
        if not _needs_creator(idx):
            continue
        for candidate in (title_ids[idx], detail.get('appId'), detail.get('agentIdentityId')):
            if candidate:
                unresolved_keys.add(str(candidate))
    audit_map = _audit_creator_map(unresolved_keys)
    if audit_map:
        for idx, detail in enumerate(details):
            if not _needs_creator(idx):
                continue
            for candidate in (title_ids[idx], detail.get('appId'), detail.get('agentIdentityId')):
                upn = audit_map.get(str(candidate or '').lower(), '')
                if upn:
                    creator_upn[idx], creator_source[idx] = upn, 'auditLog'
                    break
    print(f'tier3: resolved {len(creator_upn) - before} more')

# Remember the outcome per agent, so the next incremental run can skip the tiers.
# Agents written list-only ('missing (...)') have no cache entry and are retried next run.
for idx, info in enumerate(detail_meta):
    entry = new_detail_cache.get(info['packageId'])
    if entry is not None and idx not in _creator_settled:
        entry['creatorUpn'] = creator_upn.get(idx, '')
        entry['creatorSource'] = creator_source.get(idx, 'unattributed')

resolved = len(creator_upn)
print(f'\ncreator resolution: {resolved}/{len(details)} '
      f'({resolved / len(details) * 100:.1f}%)' if details else 'no agents')
for tier in ('ownerId', 'servicePrincipalOwner', 'auditLog'):
    n = sum(1 for v in creator_source.values() if v == tier)
    if n:
        print(f'   {tier:24} {n}')
unattributed = len(details) - resolved
if unattributed:
    print(f'   {"UNATTRIBUTED":24} {unattributed}  <- left-join these, do not drop them')

# === OWNER ACCOUNT ============================================================
owner_account = {}   # index in `details` -> Active / Disabled / Not found
if CHECK_OWNER_ACCOUNT:
    owner_keys = {
        idx: _owner_key(detail.get('ownerId') or '', creator_upn.get(idx, ''),
                        creator_source.get(idx, ''))
        for idx, detail in enumerate(details)
    }
    account_status = _owner_account_status(owner_keys.values())
    for idx, key in owner_keys.items():
        if account_status.get(key):
            owner_account[idx] = account_status[key]
    print('\nowner account:')
    for state in ('Active', 'Disabled', 'Not found'):
        n = sum(1 for v in owner_account.values() if v == state)
        if n:
            print(f'   {state:24} {n}')
    print(f'   {"unknown / no owner":24} {len(details) - len(owner_account)}')
_STEP4_OK = True
````

## 5. Shape -> canonical schema + full passthrough -> Delta

The 37 `CANONICAL` columns keep their exact upstream names and mappings so the PBIT
reads them unchanged. Every raw API field is then appended under its native name,
plus the two creator columns, `Detail As Of` and the snapshot-only `Owner account` and
`Detail status`. After the snapshot write the history table is merged and the full
detail cache is saved - in that order. Step 3's checkpoints hold only genuinely fetched
detail, so a failed run never marks a change as seen.

````python
for _flag, _step in (('_STEP3_OK', 'Step 3 (agent detail)'), ('_STEP4_OK', 'Step 4 (creators)')):
    if not globals().get(_flag):
        raise RuntimeError(f'{_step} did not complete in this session - nothing has been written. '
                           'Fix the error shown under that step and rerun from step 3 (completed '
                           'detail calls are checkpointed, so it resumes).')

from pyspark.sql import functions as F


def _join(value):
    if isinstance(value, list):
        return ';'.join(
            _json.dumps(item, ensure_ascii=False) if isinstance(item, (dict, list)) else str(item)
            for item in value if item is not None
        )
    return '' if value is None else str(value)


def _access(value):
    if value is None:
        return ''
    if not isinstance(value, list):
        raise ValueError('Agent 365 access collections must be lists when present.')
    out = []
    for item in value:
        if isinstance(item, dict):
            out.append(item.get('displayName') or item.get('id') or _json.dumps(item, ensure_ascii=False))
        else:
            out.append(str(item))
    return ';'.join(out)


# elementDetails carries vendor-supplied payloads that vary across publishers.
# A single malformed entry must not discard an entire tenant snapshot, so structural
# problems are counted and skipped; the census is printed after shaping.
_ELEMENT_SKIPS = {}


def _note_element_skip(reason):
    _ELEMENT_SKIPS[reason] = _ELEMENT_SKIPS.get(reason, 0) + 1


def _elements(detail):
    raw_groups = detail.get('elementDetails')
    if raw_groups is None:
        return '', '', ''
    if not isinstance(raw_groups, list):
        _note_element_skip('elementDetails is not a list')
        return '', '', ''
    types, bots, commands = [], [], []
    for group in raw_groups:
        if not isinstance(group, dict):
            _note_element_skip('group is not an object')
            continue
        element_type = group.get('elementType')
        if element_type:
            types.append(str(element_type))
        elements = group.get('elements') or []
        if not isinstance(elements, list):
            _note_element_skip('elements is not a list')
            continue
        for element in elements:
            if not isinstance(element, dict):
                _note_element_skip('element is not an object')
                continue
            raw_definition = element.get('definition')
            if raw_definition in (None, ''):
                continue
            if isinstance(raw_definition, str):
                try:
                    definition = _json.loads(raw_definition)
                except _json.JSONDecodeError:
                    _note_element_skip('definition is not valid JSON')
                    continue
            else:
                definition = raw_definition
            if not isinstance(definition, dict):
                _note_element_skip('definition did not decode to an object')
                continue
            if definition.get('botId'):
                bots.append(str(definition['botId']))
            for command_list in definition.get('commandLists') or []:
                if not isinstance(command_list, dict):
                    _note_element_skip('commandLists entry is not an object')
                    continue
                for command in command_list.get('commands') or []:
                    if isinstance(command, dict) and command.get('title'):
                        commands.append(str(command['title']))
            for command in definition.get('commands') or []:
                if isinstance(command, dict) and command.get('title'):
                    commands.append(str(command['title']))
    return (
        ';'.join(dict.fromkeys(types)),
        ';'.join(dict.fromkeys(bots)),
        ';'.join(dict.fromkeys(commands)),
    )


def _merge_raw_passthrough(row, raw):
    """Add every raw API field to `row` without creating a case-insensitive duplicate.

    Spark resolves column names case-insensitively (spark.sql.caseSensitive=false), so a
    raw key that differs from a canonical column only by case -- 'version' vs 'Version',
    'publisher' vs 'Publisher' -- cannot coexist with it. Keeping both fails the frame
    with AMBIGUOUS_REFERENCE on select and COLUMN_ALREADY_EXISTS on the Delta write.

    An exact-name match is already mapped, so it is skipped. For a case-only clash:
      - same value      -> skip; the canonical column already carries it
      - differing value -> keep under '<key>_raw' so the passthrough guarantee holds
    """
    folded = {key.casefold(): key for key in row}
    for api_key, api_value in raw.items():
        clash = folded.get(api_key.casefold())
        if clash is None:
            row[api_key] = api_value
            folded[api_key.casefold()] = api_key
            continue
        if str(row[clash]) == str(api_value):
            continue                       # canonical column is identical; nothing lost
        alt, suffix = f'{api_key}_raw', 2
        while alt.casefold() in folded:
            alt, suffix = f'{api_key}_raw{suffix}', suffix + 1
        row[alt] = api_value
        folded[alt.casefold()] = alt
    return row


rows = []
for idx, detail in enumerate(details):
    element_types, bot_ids, commands = _elements(detail)
    row = {
        # --- canonical (unchanged names/mappings, PBIT-compatible) ------------
        'Agent name':         detail.get('displayName') or '',
        'Title ID':           title_ids[idx],
        'Version':            detail.get('version') or '',
        'Entra Agent ID':     detail.get('agentIdentityId') or '',
        'Bot Id':             bot_ids,
        'App Id':             detail.get('appId') or '',
        'Asset Id':           detail.get('assetId') or '',
        'Publisher':          detail.get('publisher') or '',
        'Agent creator':      detail.get('publisher') or '',
        'Agent creator ID':   detail.get('ownerId') or '',
        'Agent type (A365)':  detail.get('type') or '',
        'Created in':         detail.get('platform') or '',
        'Date created':       detail.get('createdDateTime') or '',
        'Last updated':       detail.get('lastModifiedDateTime') or '',
        'Agent description':  detail.get('shortDescription') or detail.get('longDescription') or '',
        'Categories':         _join(detail.get('categories')),
        'Supported in':       _join(detail.get('supportedHosts')),
        'Availability':       detail.get('availableTo') or '',
        'Status':             detail.get('deployedTo') or '',
        'Is Blocked':         str(detail.get('isBlocked', '') or ''),
        'Users shared':       _access(detail.get('sharedWithUsersAndGroups')),
        'Groups shared':      _access(detail.get('allowedUsersAndGroups')),
        'Element types':      element_types or _join(detail.get('elementTypes')),
        'Custom actions':     commands,
        'Active Users':       str(detail.get('activeUsers', '') or ''),
        'Total sessions':     str(detail.get('totalSessions', '') or ''),
        'Exception rate':     str(detail.get('exceptionRate', '') or ''),
        'Run Time':           str(detail.get('totalRunTimeInHours', '') or ''),
        'Last Activity Date': detail.get('lastUsedDateTime') or '',

        # --- NEW: joinable creator identity ----------------------------------
        'Agent creator UPN':    creator_upn.get(idx, ''),
        'Agent creator source': creator_source.get(idx, 'unattributed'),
        # --- NEW: governance - is the owner still an active account? ---------
        'Owner account':        owner_account.get(idx, ''),
        # --- NEW: when this row's detail fields were fetched (incremental pull) --
        'Detail As Of':         detail_meta[idx]['detailAsOf'],
        # --- NEW: fetched / cached / cached - refetch failed / missing (<reason>)
        'Detail status':        detail_meta[idx].get('detailStatus', ''),
    }

    # --- NEW: full raw passthrough -------------------------------------------
    if INCLUDE_RAW_PASSTHROUGH:
        _merge_raw_passthrough(row, _flatten_raw(detail))

    rows.append(row)


def _normalise_registry_key(value):
    return (value or '').strip().upper()


def _dedupe_registry_rows(rows):
    seen, deduped = {}, []
    for row in rows:
        key = _normalise_registry_key(row.get('Title ID'))
        if not key:
            raise ValueError('Agent 365 row is missing Title ID; refusing to write an ambiguous snapshot.')
        comparable = {c: '' if v is None else str(v).strip() for c, v in row.items()}
        prior = seen.get(key)
        if prior is None:
            seen[key] = comparable
            deduped.append(row)
            continue
        if comparable != prior:
            raise ValueError(f'Conflicting Agent 365 rows detected for {key!r}; refusing to overwrite a good snapshot.')
    return deduped


def _guard_registry_snapshot(rows, table_name):
    exists = bool(spark.catalog.tableExists(table_name))
    if not rows and exists:
        raise ValueError(f'Fetched 0 Agent 365 rows; refusing to replace existing {table_name}.')
    if not rows and not ALLOW_EMPTY_SNAPSHOT:
        raise ValueError(
            f'Fetched 0 Agent 365 rows and {table_name} does not exist yet. '
            'Set ALLOW_EMPTY_SNAPSHOT = True only for an intentional empty first install.'
        )


rows = _dedupe_registry_rows(rows)
_guard_registry_snapshot(rows, TARGET_TABLE)

rows = [{k: ('' if v is None else str(v)) for k, v in row.items()} for row in rows]

CANONICAL = [
    'Agent name', 'Supported in', 'Date created', 'Agent creator', 'Publisher',
    'Agent type (A365)', 'Version', 'Availability', 'Agent creator ID',
    'Agent description', 'Created in', 'Last updated', 'Custom actions',
    'Title ID', 'Sensitivity',
    'Can read OneDrive and Sharepoint items', 'OneDrive and Sharepoint items',
    'Can read OneDrive files', 'OneDrive files', 'OneDrive sites',
    'Can read Sharepoint sites and files', 'Sharepoint files', 'Sharepoint sites',
    'Can extend to Graph connector', 'Graph connector details',
    'Can generate images using user prompt', 'Can use code interpreter',
    'Contains uploaded files', 'Uploaded files', 'Status',
    'Active Users', 'Total sessions', 'Exception rate', 'Last Activity Date',
    'Deployment', 'Run Time', 'Risks',
]
NEW_CANONICAL = ['Agent creator UPN', 'Agent creator source', 'Detail As Of']
# Current state only: kept out of the history table, whose schema predates them.
SNAPSHOT_ONLY = ['Owner account', 'Detail status']

# Union every key across every row so schema is stable even when the API returns
# different attributes per agent.
all_keys = []
for row in rows:
    for k in row:
        if k not in all_keys:
            all_keys.append(k)
for column in CANONICAL + NEW_CANONICAL + SNAPSHOT_ONLY:
    if column not in all_keys:
        all_keys.append(column)
for row in rows:
    for column in all_keys:
        row.setdefault(column, '')

_named = CANONICAL + NEW_CANONICAL + SNAPSHOT_ONLY
ordered = _named + [c for c in all_keys if c not in _named]

if rows:
    df = spark.createDataFrame(rows).select(*[F.col(f'`{c}`') for c in ordered])
else:
    df = spark.createDataFrame([], ','.join(f'`{c}` string' for c in ordered))

filled = {c: sum(1 for r in rows if r.get(c)) for c in ordered}
print(f'packages shaped : {len(rows)}')
print(f'total columns   : {len(ordered)}  '
      f'(canonical {len(CANONICAL)} + new {len(NEW_CANONICAL) + len(SNAPSHOT_ONLY)} + passthrough '
      f'{len(ordered) - len(_named)})')
print('\npopulated columns:')
for c in ordered:
    if filled.get(c):
        print(f'   {c:44} {filled[c]}/{len(rows)}')
blank = [c for c in ordered if not filled.get(c)]
if blank:
    print(f'\nblank ({len(blank)}): {", ".join(blank)}')
    print('   -> observability-only fields are populated by the Admin Center CSV export only.')
if _ELEMENT_SKIPS:
    print(f'\nelementDetails entries skipped ({sum(_ELEMENT_SKIPS.values())}):')
    for _reason in sorted(_ELEMENT_SKIPS):
        print(f'   {_reason:44} {_ELEMENT_SKIPS[_reason]}')
    print('   -> malformed publisher payloads; set INCLUDE_RAW_PASSTHROUGH to retain them verbatim.')

(df.write.mode(WRITE_MODE)
   .option('overwriteSchema', 'true')
   .option('delta.columnMapping.mode', 'name')
   .option('delta.minReaderVersion', '2')
   .option('delta.minWriterVersion', '5')
   .format('delta').saveAsTable(TARGET_TABLE))
print(f'\nwrote -> {TARGET_TABLE} ({WRITE_MODE}) | columns: {len(df.columns)}')


HISTORY_KEY = ('Title ID', 'Last updated')


def _history_merge_clauses(columns):
    """Build the MERGE condition and SET map for the registry history table.

    One row per (Title ID, Last updated): an unchanged agent updates its row in place
    (usage fields, Detail As Of, Last Seen), an edited agent gets a new row. Column
    names carry spaces, so every reference is back-quoted.
    """
    condition = ' AND '.join(f't.`{c}` = s.`{c}`' for c in HISTORY_KEY)
    updates = {f'`{c}`': f's.`{c}`' for c in columns
               if c not in HISTORY_KEY and c != 'First Seen'}
    return condition, updates


def _write_registry_history(snapshot_df, table_name, run_as_of):
    if not table_name:
        return
    columns = CANONICAL + NEW_CANONICAL + ['First Seen', 'Last Seen']
    if 'Detail status' in snapshot_df.columns:
        # List-only rows carry no detail; merging them would blank the usage fields of
        # the agent's history row. They reach history once their detail is fetched.
        snapshot_df = snapshot_df.filter(~F.col('`Detail status`').startswith('missing'))
    source = (snapshot_df.select(*[F.col(f'`{c}`') for c in CANONICAL + NEW_CANONICAL])
                         .withColumn('First Seen', F.lit(run_as_of))
                         .withColumn('Last Seen', F.lit(run_as_of))
                         .dropDuplicates(list(HISTORY_KEY)))
    if not spark.catalog.tableExists(table_name):
        (source.write.mode('overwrite')
               .option('delta.columnMapping.mode', 'name')
               .option('delta.minReaderVersion', '2')
               .option('delta.minWriterVersion', '5')
               .format('delta').saveAsTable(table_name))
        print(f'history  -> {table_name} created ({source.count()} rows)')
        return
    from delta.tables import DeltaTable
    condition, updates = _history_merge_clauses(columns)
    (DeltaTable.forName(spark, table_name).alias('t')
        .merge(source.alias('s'), condition)
        .whenMatchedUpdate(set=updates)
        .whenNotMatchedInsertAll()
        .execute())
    print(f'history  -> {table_name} merged on {" + ".join(HISTORY_KEY)}')


def _write_detail_cache(cache, table_name):
    if not table_name:
        return
    schema = ('package_id string, last_modified string, detail_as_of string, '
              'creator_upn string, creator_source string, detail_json string')
    cache_df = spark.createDataFrame(_cache_to_rows(cache), schema)
    (cache_df.write.mode('overwrite').option('overwriteSchema', 'true')
             .format('delta').saveAsTable(table_name))
    print(f'cache    -> {table_name} ({len(cache)} agents)')


# The snapshot is written; only now record history and the full detail cache. Step 3
# checkpoints only genuinely fetched detail, so a run that failed earlier never marks a
# change as seen; this final write also prunes removed agents and records creators.
_write_registry_history(df, HISTORY_TABLE, RUN_AS_OF)
_write_detail_cache(new_detail_cache, DETAIL_CACHE_TABLE)
````

## 6. Downstream wiring

**Lakehouse join** - `Agent creator UPN` is already lowercased/trimmed, so it matches
`PersonId_Normalized` directly:

```sql
SELECT a.[Title ID], a.[Agent name], a.[Agent creator source],
       COALESCE(o.Level3_Name, 'Unattributed') AS Franchise,
       o.JobTitle
FROM   dbo.agents_365 a
LEFT   JOIN dbo.copilot_org_data o
       ON a.[Agent creator UPN] = o.PersonId_Normalized;
```

**LEFT join, always.** An inner join silently drops every unattributed agent and
understates your totals.

**Power BI model** - relationship `agents_365[Agent creator UPN]` -> `org[PersonId_Normalized]`,
many-to-one, single direction. Set `Agent creator source` as a slicer so coverage is
visible on the page rather than assumed.

**Franchise from Level1-14** - these are positional, not semantic. `Level3_Name` is a
placeholder above: a person six hops from the CEO and one eleven hops down have their
franchise at different levels. Prefer Graph `companyName`/`department`, or build a
`franchise_lookup` mapping table and `COALESCE` across levels to find the first match.
